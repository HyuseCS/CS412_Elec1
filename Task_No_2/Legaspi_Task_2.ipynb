# Descriptive Statistics + Plots

Random data from NumPy. Find mean, mode, median, variance, standard deviation. Then plot it.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=10, size=200).round()   # rounded so a mode exists
data[:10]
```

## Statistics

```python
print("Mean    :", np.mean(data))
print("Median  :", np.median(data))
print("Mode    :", stats.mode(data, keepdims=False).mode)
print("Variance:", np.var(data, ddof=1))
print("Std dev :", np.std(data, ddof=1))
```

## Plots

```python
fig, ax = plt.subplots(2, 2, figsize=(11, 8))

ax[0, 0].hist(data, bins=20, edgecolor="black")
ax[0, 0].set_title("Histogram")

ax[0, 1].boxplot(data)
ax[0, 1].set_title("Box plot")

stats.probplot(data, plot=ax[1, 0])
ax[1, 0].set_title("Q-Q plot")

ax[1, 1].scatter(np.arange(data.size), data, s=12)
ax[1, 1].set_title("Scatter (value vs index)")

plt.tight_layout()
plt.show()
```
